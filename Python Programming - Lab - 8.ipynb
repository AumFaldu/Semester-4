### 02) Write a function that add first n numbers.

```python
n = int(input("Enter value of n : "))
def add(n):
    sum=0
    for i in range(1,n+1):
        sum += i
    return sum
print(f"Sum of first {n} numbers = {add(n)}")
```

### 03) Write a function that returns 1 if the given number is Prime or 0 otherwise.

```python
n = int(input("Enter value of n : "))
def isPrime(n):
    for i in range(2,int(n/2)+1):
        if n%i==0:
            return 0
    else:
        return 1
print("",isPrime(n))
```

### 04) Write a function that returns the list of Prime numbers between given two numbers.

```python
n1 = int(input("Enter smallest number : "))
n2 = int(input("Enter largest number : "))
def range_prime(n1,n2):
    li = []
    for i in range(n1,n2):
        flag = True
        for j in range(2,int(i/2)+1):
            if i%j==0:
                flag=False
                break
        if flag==True:
            li.append(i)
    return li
print(f"List of Prime numbers = {range_prime(n1,n2)}")
```

### 08) Write a recursive function to find nth term of Fibonacci Series.

```python
n = int(input("Enter value of n : "))
def Fibbo(n):
    if n==0 or n==1:
        return n
    else:
        return Fibbo(n-1)+Fibbo(n-2)
print(f"Term {n} = ",Fibbo(n))
```

### 11) Write a function to invert a given Dictionary.
#### hint: keys to values & values to keys
#### Before : {'a': 10, 'b':20, 'c':30, 'd':40}
#### After : {10:'a', 20:'b', 30:'c', 40:'d'}

```python
keys = input("Enter space separated keys : ").split()
values = list(map(int,input("Enter space separated values : ").split()))
d = {i:j for i,j in zip(keys,values)}
def invert(d):
    return {j:i for i,j in d.items()}
print("Invert = ",invert(d))
```

### 12) Write a function to check whether the given string is Pangram or not.
#### hint: Pangram is a string containing all the characters a-z atlest once.
#### "the quick brown fox jumps over the lazy dog" is a Pangram string.

```python
import string

def isPangram(s):
    alphabet = set(string.ascii_lowercase)
    return set(s.lower()) >= alphabet

s = input("Enter a string: ")
print(isPangram(s))
```

### 13) Write a function that returns the number of uppercase and lowercase letters in the given string. 
#### example : Input : s1 = AbcDEfgh ,Ouptput : no_upper = 3, no_lower = 5

```python
s = input("Enter a string : ")
def count_upper_lower(s):
    no_upper=0
    no_lower=0
    for i in s:
        if i.isupper():
           no_upper += 1
        elif i.islower():
            no_lower += 1
    print(f"Upper = {no_upper} \n Lower = {no_lower}")
count_upper_lower(s)
```

### 15) For the given list of names of students, extract the names having more that 7 characters. Use filter().

```python
names = input("Enter space separated names : ")
def greater_than_7(names):
    names = names.split()
    return list(filter(lambda i:len(i)>7,names))
print("Names having length greater than 7 = ",greater_than_7(names))
```

### 16) For the given list of names of students, convert the first letter of all the names into uppercase. use map().

```python
names = input("Enter space separated names : ")
def first_capital(names):
    names = names.split()
    return " ".join(list(map(lambda x:x[0].upper()+x[1:],names)))
print("Capitalize first letter = ",first_capital(names))
```
