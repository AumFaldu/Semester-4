### 04) WAP to demonstrate the use of Pie chart.

```python
labels = ['Apple', 'Banana', 'Cherry', 'Dates'] 
sizes = [25, 35, 20, 20] 
colors = ['red', 'yellow', 'pink', 'brown']
plt.pie(sizes,labels=labels,colors=colors,autopct='%1.1f%%')
plt.title("Fruit Distribution")
plt.show()
```

### 05) WAP to demonstrate the use of Bar chart.

```python
categories = ['Apple', 'Banana', 'Cherry', 'Dates']
values = [25, 35, 20, 40]
plt.bar(categories,values,color='blue')
plt.xlabel('Fruits')
plt.ylabel('Quantity')
plt.title('Fruit Distribution')
plt.show()
```

### 06) WAP to demonstrate the use of Scatter Plot.

```python
x = [1, 2, 3, 4, 5]
y = [10, 20, 25, 30, 35]
sizes = [50, 100, 200, 300, 400]
colors = ['red', 'blue', 'green', 'orange', 'purple']
plt.scatter(x,y,s=sizes,c=colors,alpha=0.7,edgecolors='black')
plt.xlabel('X-axis Label')
plt.ylabel('Y-axis Label')
plt.title('Scatter Plot')
plt.show()
```

### 08) WAP to display the value of each bar in a bar chart using Matplotlib.

```python
categories = ['A', 'B', 'C', 'D', 'E']
values = [23, 45, 12, 67, 34]
plt.bar(categories,values,color='blue',edgecolor='black')
for i, value in enumerate(values):
    plt.text(i, value + 1, str(value), ha='center', va='bottom', fontsize=10, color='black')
plt.xlabel('Categories')
plt.ylabel('Values')
plt.title('Bar Chart with Values')
plt.show()
```

### 09) WAP create a Scatter Plot with several colors in Matplotlib?

```python
import random
x = [random.uniform(0, 10) for _ in range(50)] 
y = [random.uniform(0, 10) for _ in range(50)] 

colors = [random.uniform(0, 1) for _ in range(50)]
sizes = [random.randint(10, 100) for _ in range(50)]

plt.scatter(x, y, c=colors, s=sizes, cmap='plasma', alpha=0.8, edgecolor='black')

plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.title('Scatter Plot with Several Colors (No NumPy)')

plt.colorbar(label='Color Intensity')

plt.show()
```

### 10) WAP to create a Box Plot.

```python
import random

data1 = [random.randint(10, 50) for _ in range(20)]
data2 = [random.randint(30, 70) for _ in range(20)]
data3 = [random.randint(20, 90) for _ in range(20)]

plt.boxplot([data1, data2, data3], patch_artist=True,
            boxprops=dict(facecolor='lightblue', color='blue'),
            whiskerprops=dict(color='blue'),
            capprops=dict(color='blue'),
            medianprops=dict(color='red'))

plt.xticks([1, 2, 3], ['Dataset 1', 'Dataset 2', 'Dataset 3']) 
plt.xlabel('Datasets')
plt.ylabel('Values')
plt.title('Box Plot Example')

plt.show()
```
