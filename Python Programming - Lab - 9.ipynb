### 08) WAP to create an UDF named frequency to count occurances of the specific word in a given text file.

```python
def frequency(FileName, word):
    with open(FileName, "r") as fp:
        lines = fp.readlines()
        words = [w for line in lines for w in line.strip().split()]  # Flatten the list of words
    print(f"Occurrence of '{word}' = {words.count(word)}")

FileName = input("Enter File Name: ")
word = input("Enter word: ")
frequency(FileName, word)
```

### 10) WAP to write first 100 prime numbers to a file named primenumbers.txt 
(Note: each number should be in new line)

```python
def is_prime(n):
    if n < 2:
        return False
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False
    return True

def write_primes_to_file(filename, count):
    primes = []
    num = 2
    while len(primes) < count:
        if is_prime(num):
            primes.append(num)
        num += 1

    with open(filename, "w") as fp:
        fp.writelines([str(prime) + "\n" for prime in primes])

write_primes_to_file("primenumbers.txt", 100)
```
