# Exception Handling

### 01) WAP to handle following exceptions:
1. ZeroDivisionError
2. ValueError
3. TypeError
#### Note: handle them using separate except blocks and also using single except block too.

```python
#Separate Except blocks
#ZeroDivisionError
try:
    print(1/0)
except ZeroDivisionError as err:
    print(err)
#ValueError
try:
    a = int(input("Enter a string : "))
    print(a)
except ValueError as err:
    print(err)
#TypeError
try:
    s = input("Enter string : ")
    print(s+1)
except TypeError as err:
    print(err)
#Single Except block
try:
    choice = int(input("Enter 1 for ZeroDivisionError\nEnter 2 for ValueError\nEnter 3 for TypeError"))
    match(choice):
        case 1:
            print(1/0)
        case 2:
            int("abc")
        case 3:
            "a"+1
except (ZeroDivisionError,ValueError,TypeError) as err:
    print(err)
```

### 08) WAP that gets an age of a person form the user and raises ValueError with error message: "Enter Valid Age" :
#### If the age is less than 18. 
#### otherwise print the age.

```python
try:
    age = int(input("Enter age : "))
    if age<18:
        raise ValueError("Error:Enter Valid Age")
    else:
        print(age)
except ValueError as err:
    print(err)
```

### 09) WAP to raise your custom Exception named InvalidUsernameError with the error message : "Username must be between 5 to 15 characters long":
#### if the given name is having characters less than 5 or greater than 15.
#### otherwise print the given username.

```python
class InvalidUsernameError(Exception):
    def __init__(self,msg):
        self.msg=msg
try:
    username = input("Enter Username : ")
    if len(username)<5 or len(username)>15:
        raise InvalidUsernameError("Username must be between 5 to 15 characters long")
    else:
        print(username)
except InvalidUsernameError as err:
    print(err)
```

### 10) WAP to raise your custom Exception named NegativeNumberError with the error message : "Cannot calculate the square root of a negative number" :
#### if the given number is negative.
#### otherwise print the square root of the given number.

```python
class NegativeNumberError(Exception):
    def __init__(self,msg):
        self.msg=msg
try:
    number = int(input("Enter a number : "))
    if number<0:
        raise NegativeNumberError("Cannot calculate the square root of a negative number")
    else:
        print(number**0.5)
except NegativeNumberError as err:
    print(err)
```
