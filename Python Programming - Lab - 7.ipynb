# Set & Dictionary

### 01) WAP to iterate over a set.

```python
n = input("Enter space separated values : ")
s = set(map(int,n.split()))
for i in s:
    print(i)
```

### 02) WAP to convert set into list, string and tuple.

```python
n = input("Enter space separated values : ")
s = set(map(int,n.split()))
li = list(s)
st = str(s)
t = tuple(s)
print("List =",li)
print("String =",st)
print("Tuple =",t)
```

### 03) WAP to find Maximum and Minimum from a set.

```python
n = input("Enter space separated values : ")
s = set(map(int,n.split()))
print("Maximum = ",max(s))
print("Minimum = ",min(s))
```

### 04) WAP to perform union of two sets.

```python
n1 = input("Enter space separated values : ")
n2 = input("Enter space separated values : ")
s1 = set(map(int,n1.split()))
s2 = set(map(int,n2.split()))
print("Union using union() = ",s1.union(s2))
print("Union using | = ",s1|s2)
```

### 05) WAP to check if two lists have at-least one element common.

```python
l1 = input("Enter space separated values : ").split()
l2 = input("Enter space.separated values : ").split()
s1 = set(map(int,l1))
s2 = set(map(int,l2))
if s1.intersection(s2):
    print("Atleast one element is common")
else:
    print("No element is common")
```

### 09) WAP to check whether two given strings are anagram using dictionary.

```python
s1 = input("Enter a string ")
s2 = input("Enter a string ")
if len(s1)!=len(s2):
    print("Not Anagrams")
else:
    count1 = {}
    count2 = {}
    for i in s1:
        count1[i] = s1.count(i)
    for i in s2:
        count2[i] = s2.count(i)
    if count1==count2:
        print("Both are anagrams")
    else:
        print("Not Anagrams")
```

### 10) WAP to find common elements in three lists using set.

### 12) WAP to check if a given string is binary string or not.

```python
n = input("Enter a string:")
s = set("01")
for i in n:
    if i not in s:
        print("Not a Binary String")
        break
else:
    print("Is a Binary String")
```

### 13) WAP to sort dictionary by key or value.

```python
key = input("Enter keys in space-separated form: ").split()
key = map(int, key)
value = input("Enter values in space-separated form: ").split()
value = map(int, value)
d = {i: j for (i, j) in zip(key, value)}
sorted_by_key = {k: v for k, v in sorted(d.items())}
sorted_by_value = {k: v for k, v in sorted(d.items(), key=lambda item: item[1])}
print("Sort by Key =", sorted_by_key)
print("Sort by Value =", sorted_by_value)
```

### 14) WAP to find the sum of all items (values) in a dictionary given by user. (Assume: values are numeric)

```python
key = input("Enter keys in space-separated form: ").split()
key = map(int, key)
value = input("Enter values in space-separated form: ").split()
value = map(int, value)
d = {i: j for (i, j) in zip(key, value)}
print("Sum of values = ",sum(d.values()))
```

### 15) WAP to handle missing keys in dictionaries.
#### Example : Given, dict1 = {'a': 5, 'c': 8, 'e': 2} 
#### if you look for key = 'd', the message given should be 'Key Not Found', otherwise print the value of 'd' in dict1.

```python
key = input("Enter keys in space-separated form: ").split()
value = input("Enter values in space-separated form: ").split()
value = map(int, value)
dict1 = {i:j for (i,j) in zip(key,value)}
n = input("Enter the key : ")
if n not in dict1.keys():
    print("Key Not Found")
else:
    print("Value = ",dict1[n])
```
